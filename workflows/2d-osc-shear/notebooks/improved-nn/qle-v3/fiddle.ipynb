```python
evecs = data["evecs"]
evals = data["evals"]
```

```python
out = []
for i in range(len(evals)):
    term = np.linalg.norm(evecs[i])**2/evals[i]**2
    out.append(term)
```

```python
iprs = []
for i in range(len(evals)):
    norm = np.linalg.norm(evecs[i], axis=-1)
    ipr = np.sum(norm**4)/np.sum(norm**2)**2
    iprs.append(ipr)
```

```python
plt.plot(iprs)
# plt.xlim(0, 512)
# plt.yscale("log")
```
